```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import warnings
warnings.filterwarnings("ignore")

# Dataset
data = {
    "Weight":[50,60,70,80,90],
    "Calories":[1800,2000,2200,2500,2800],
    "Walk_km":[3,4,5,6,7],
    "Water_liters":[2.5,3,3.5,4,4.5]
}

df = pd.DataFrame(data)

# Input and output
X = df[["Weight"]]
y = df[["Calories","Walk_km","Water_liters"]]

# Model
model = LinearRegression()
model.fit(X,y)

# User input
weight = float(input("Enter your weight (kg): "))

# Prediction
prediction = model.predict([[weight]])

print("\nDiet Recommendation")
print("Calories per day:", round(prediction[0][0]))
print("Walking distance:", round(prediction[0][1]),"km")
print("Water intake:", round(prediction[0][2],2),"liters")
```

Problem:
People don't know how much calories, water, and exercise they need.

Solution:
Machine Learning model predicts diet plan based on weight.

Algorithm:
Linear Regression
