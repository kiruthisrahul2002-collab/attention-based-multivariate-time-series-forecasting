```python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor

# Dataset
data = {
    "weight":[50,80,120,0.8,1.2,1.5,2.0],
    "weight_unit":["g","g","g","kg","kg","kg","kg"],
    "age":[5,10,20,1,2,4,6],
    "age_unit":["days","days","days","months","months","months","months"],
    "feed":[20,25,35,80,100,120,150]
}

df = pd.DataFrame(data)

# Convert weight to kg
for i in range(len(df)):
    if df.loc[i,"weight_unit"] == "g":
        df.loc[i,"weight"] = df.loc[i,"weight"] / 1000

# Convert age to months
for i in range(len(df)):
    if df.loc[i,"age_unit"] == "days":
        df.loc[i,"age"] = df.loc[i,"age"] / 30

X = df[["weight","age"]]
y = df["feed"]

model = RandomForestRegressor(n_estimators=100,random_state=42)
model.fit(X,y)

total_daily_feed = 0

while True:

    hens = int(input("Enter number of hens: "))

    weight_unit = input("Enter weight unit (kg/g): ").lower()
    weight = float(input("Enter hen weight: "))

    if weight_unit == "g":
        weight = weight / 1000

    age_unit = input("Enter age unit (days/months): ").lower()
    age_value = float(input("Enter age value: "))

    if age_unit == "days":
        months = age_value / 30
    else:
        months = age_value

    # ML prediction
    feed_per_hen = model.predict([[weight,months]])[0]

    daily_feed = (feed_per_hen * hens) / 1000

    total_daily_feed += daily_feed

    print("\nFeed per hen:",round(feed_per_hen,2),"grams")
    print("Batch daily feed:",round(daily_feed,2),"kg")

    choice = input("\nAdd another batch? (yes/stop): ").lower()

    if choice == "stop":

        total_monthly_feed = total_daily_feed * 30

        print("\n------ Total Feed Requirement ------")
        print("Total feed per day:",round(total_daily_feed,2),"kg")
        print("Total feed per month:",round(total_monthly_feed,2),"kg")

        print("Program stopped.")
        break
```

Project Title:
AI-Based Poultry Feed Planning System Using Random Forest

Project Explanation:

In this project, I developed a machine learning-based system to help poultry farmers estimate the required amount of feed for their chickens.

In many poultry farms, farmers have different groups of chickens such as chicks, medium-sized birds, and adult hens. Each group requires a different amount of feed depending on their weight and age. However, farmers often calculate feed requirements manually, which can lead to feed wastage or shortages.

To solve this problem, I built a system using the Random Forest Regression algorithm that predicts the daily feed requirement per hen based on two main features: the weight of the hen and its age.

The system accepts user inputs such as:

Number of hens

Hen weight (with unit options in kilograms or grams)

Age of the hen (with options in days, weeks, or months)

The program automatically converts these units into standard formats before passing them to the machine learning model.

After predicting the feed required per hen, the system calculates:

Total feed required for that batch per day

A feeding schedule divided into morning and evening feeding

Weekly and monthly feed requirements

The system also supports multiple poultry batches, which allows farmers to enter different groups such as chicks, medium hens, and adult hens. When the user stops the program, it calculates the total feed required per day and per month for the entire farm.
